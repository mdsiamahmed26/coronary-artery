```python
# Install VTK
!pip install vtk

import vtk
import numpy as np
from google.colab import files

# === Step 1. Upload your artery centerline file (.vtp) ===
print("Please upload your .vtp centerline file (extracted from artery)")
uploaded = files.upload()
filename = list(uploaded.keys())[0]

# === Step 2. Read centerline ===
reader = vtk.vtkXMLPolyDataReader()
reader.SetFileName(filename)
reader.Update()
centerline = reader.GetOutput()

# === Step 3. Compute curvature ===
curvature_filter = vtk.vtkCurvatures()
curvature_filter.SetInputData(centerline)
curvature_filter.SetCurvatureTypeToMaximum()  # Options: Maximum, Minimum, Gaussian, Mean
curvature_filter.Update()
curved_data = curvature_filter.GetOutput()

# Extract curvature values
curvature_array = curved_data.GetPointData().GetScalars()
curvatures = np.array([curvature_array.GetValue(i) for i in range(curvature_array.GetNumberOfTuples())])

# === Step 4. Compute numerical metrics ===
mean_curvature = np.mean(curvatures)
max_curvature = np.max(curvatures)
min_curvature = np.min(curvatures)

print("=== Artery Centerline Curvature Metrics ===")
print(f"Mean curvature (κ_mean): {mean_curvature:.4f} mm^-1")
print(f"Maximum curvature (κ_max): {max_curvature:.4f} mm^-1")
print(f"Minimum curvature (κ_min): {min_curvature:.4f} mm^-1")
```
